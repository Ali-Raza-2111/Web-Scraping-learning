```python
import pandas as pd
import numpy as np

# Sample DataFrame for all questions
data = {
    'Product': ['Laptop', 'Phone', 'Tablet', 'Monitor', 'Keyboard', 'Mouse', 'Printer', 'Camera'],
    'Price': [1200, 800, 450, 300, 50, 25, 200, 600],
    'Category': ['Electronics', 'Electronics', 'Electronics', 'Electronics', 
                'Accessory', 'Accessory', 'Office', 'Electronics'],
    'Stock': [15, 30, 0, 12, 45, 60, 8, 5],
    'Rating': [4.5, 4.2, 3.9, 4.0, 3.5, 4.1, 3.8, 4.7]
}
df = pd.DataFrame(data)

# Question 1: Basic numeric filter - select products with Price > 500
# Your code here
print(df.query('Price > 500'))
print('====================')
# Question 2: String filter - select Electronics products
# Your code here
print(df.query('Category == "Electronics"'))
print('====================')
# Question 3: Multiple conditions - select Electronics with Price < 1000
# Your code here
print(df.query('Category == "Electronics" and Price < 1000'))
print('====================')
# Question 4: Using OR condition - select products with Price < 100 OR Price > 800
# Your code here
print(df.query("100 < Price > 800"))
print('====================')

# Question 5: Filter with external variable - select products cheaper than max_price
max_price = 400
# Your code here
print(df.query('Price < @max_price'))
print('==========================================================')

# Question 6: String operation - select products containing 'o' in their name
# Your code here
print(df.query('Product.str.contains("o")'))
print('==========================================================')

# Question 7: Null check - select products that are out of stock (Stock == 0)
# Your code here
print(df.query('Stock == 0'))
print('==========================================================')
# Question 8: Math operation - select products where Price is at least 10×Stock
# Your code here
print(df.query('Price < 10*Stock'))
print('==========================================================')
# Question 9: Index query - select products where index is even (0, 2, 4...)
# Your code here
print(df.query('index%2 == 0'))
print('==========================================================')
# Question 10: Complex condition - select Electronics with Rating > 4 OR 
# Accessories with Price < 75
print(df.query('Category == "Electronics" and (Rating > 4 or  Price < 75)'))
# Your code here
```
