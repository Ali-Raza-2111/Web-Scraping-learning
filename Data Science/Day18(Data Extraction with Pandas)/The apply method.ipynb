```python
import pandas as pd

inventory = pd.DataFrame({
    'Item': ['Apples', 'Milk', 'Eggs', 'Bread', 'Cheese', 'Chicken'],
    'Category': ['Produce', 'Dairy', 'Dairy', 'Bakery', 'Dairy', 'Meat'],
    'Price': [2.99, 3.49, 4.19, 2.49, 5.99, 8.99],
    'Quantity': [150, 80, 120, 95, 60, 45],
    'Perishable': [True, True, True, False, True, True]
})
```

```python
def make_price_label(price):
    if price <3:
        return "Budget"
    elif price<6:
        return "Standard"
    else:
        return "Premium"
    
inventory['Price_Label'] = inventory['Price'].apply(make_price_label)
inventory
```

```python
def  restock_check(quantity):
    if quantity < 50:
        return "Urgent"
    elif quantity < 100:
        return "Monitor"
    else :
        return "Adequate"
    
inventory['Stock_Status'] = inventory['Quantity'].apply(restock_check)
inventory
            
```

```python
def describe_item(row):
    return f"{row['Item']} {row['Category']} - {row['Price']}"

inventory['Description'] = inventory.apply(describe_item,axis= 1)
inventory
```

```python
def storage_needs(row):
    if row['Perishable']:
        return 'Refrigerate'
    elif (row['Perishable'] == False) and (row['Category'] == 'Bakery'):
        return 'Shelf'
    else:
        return "Dry Storage"
    
inventory['Storage'] = inventory.apply(storage_needs,axis=1)
inventory
```

```python
def order_priority(row):
    if (row['Price'] > 5) and (row['Quantity'] < 75):
        return 'High'
    elif  (row['Price'] > 3) or (row['Quantity'] < 100):
        return "Medium"
    else:
        return 'Low'
    
inventory['Priority'] = inventory.apply(order_priority,axis = 1)
inventory
```
