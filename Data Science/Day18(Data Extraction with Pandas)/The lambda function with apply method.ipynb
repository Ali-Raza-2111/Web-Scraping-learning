```python
import pandas as pd

# Practice DataFrame
sales_data = pd.DataFrame({
    'Product': ['Laptop', 'Phone', 'Tablet', 'Headphones', 'Mouse'],
    'Unit_Price': [1200, 800, 450, 150, 25],
    'Quantity_Sold': [15, 30, 25, 40, 60],
    'Discount_Percent': [10, 5, 15, 20, 0]
})

# Question 1: Create a new column 'Final_Price' that applies the discount to Unit_Price
# using lambda (Final_Price = Unit_Price * (1 - Discount_Percent/100))
sales_data['Final_Price'] = sales_data.apply(lambda row:(row['Unit_Price'] * (1-row['Discount_Percent']/100)), axis=1)

# Question 2: Create a 'Total_Sales' column (Unit_Price * Quantity_Sold) using lambda
sales_data['Total_Sales'] = sales_data.apply(lambda row:row['Unit_Price']*row['Quantity_Sold'], axis=1)

# Question 3: Create a 'Price_Category' column that labels products as:
# "Premium" (≥$500), "Mid-Range" ($100-$499), "Budget" (<$100) using lambda
sales_data['Price_Category'] = sales_data['Unit_Price'].apply(lambda x:"Premium" if x > 500 else "Mid-Range" if 100 >x<499 else "Budget")

# Question 4: Create a 'Stock_Alert' column that shows:
# "Reorder" if Quantity_Sold > 50, else "OK" using lambda
sales_data['Stock_Alert'] = sales_data['Quantity_Sold'].apply(lambda x:"Reorder" if x > 50 else "OK")

# Question 5: Create a 'Description' column combining Product and Price_Category
# Example: "Laptop (Premium)" using lambda (use axis=1 for row-wise operation)
sales_data['Description'] = sales_data.apply(lambda row: f"{row['Product']} ({row['Price_Category']})", axis=1)

# After completing, check your results:
print(sales_data)
```
