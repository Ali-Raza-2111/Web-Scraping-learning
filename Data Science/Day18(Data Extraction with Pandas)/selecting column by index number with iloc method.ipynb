```python
import pandas as pd

# 🔹 Create a sample DataFrame
data = {
    'Student': ['Aisha', 'Ben', 'Carlos', 'Diana', 'Eva'],
    'Math': [88, 92, 79, 85, 90],
    'Science': [91, 89, 84, 87, 93],
    'English': [85, 87, 78, 90, 88]
}

df = pd.DataFrame(data)

print("📊 DataFrame:\n")
print(df)

# ========================================================
# 🟨 Q1. Get the English score of Eva using `.iloc`.
# Hint: Eva is the last row, and English is the 4th column (index 3).
# 👉 Your code here:
print(df.iloc[-1,3])

# ========================================================
# 🟨 Q2. Retrieve the names and Science scores of the first three students.
# Hint: Rows = 0 to 2, Columns = 0 (Student) and 2 (Science)
# 👉 Your code here:
print(df.iloc[0:3,[0,2]])

# ========================================================
# 🟨 Q3. Get the entire row of the student with the lowest Math score.
# Hint: Use df['Math'].idxmin() to get the row index.
# 👉 Your code here:
df.iloc[df['Math'].idxmin(),:]

# ========================================================
# 🟨 Q4. Show only the Math and English scores of Ben and Carlos.
# Hint: Ben and Carlos are rows 1 and 2. Math = column 1, English = column 3.
# 👉 Your code here:
df.iloc[[1,2],[1,3]]

# ========================================================
# 🟨 Q5. Use negative indexing to get the Science and English scores 
#       of the last two students (Diana and Eva).
# Hint: Last two rows = -2:, Science = col 2, English = col 3
# 👉 Your code here:

df.iloc[-2:,[2,3]]
```
