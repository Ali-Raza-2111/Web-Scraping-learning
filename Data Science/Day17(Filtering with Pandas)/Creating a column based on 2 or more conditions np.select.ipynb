```python

import pandas as pd
import numpy as np

# Sample DataFrame
data = {
    'Student': ['Amy', 'Brian', 'Clara', 'Dan', 'Ella'],
    'Score': [95, 83, 76, 68, 59],
    'Attendance': [98, 89, 75, 70, 65]
}
df = pd.DataFrame(data)

print("Original DataFrame:")
print(df)

# -------------------------------------
# 📌 Question 1:
# Create a column "Grade" based on the following:
# - "A" if Score >= 90
# - "B" if Score is between 80–89
# - "C" if Score is between 70–79
# - "D" if Score is between 60–69
# - "F" otherwise

conditions1 = [
    df['Score'] >= 90,
    df['Score'] >= 80,
    df['Score'] >= 70,
    df['Score'] >= 60,
    df['Score'] < 60
]
values1 = ["A","B","C","D","F"]
df["Grade"] = np.select(conditions1,values1,default='other')

# -------------------------------------
# 📌 Question 2:
# Create a column "Discipline" based on Attendance:
# - "Excellent" if Attendance >= 95
# - "Good" if Attendance is between 85–94
# - "Average" if Attendance is between 70–84
# - "Poor" otherwise

# Your code here using np.select()
conditions2 = [
    df['Attendance']>=95,
    df['Attendance']>=85,
    df['Attendance']>=70,
    df['Attendance']<70
]
values2 = ["Excellent","Good","Average","Poor"]
df["Discipline"] = np.select(conditions2,values2,default='other')
# -------------------------------------
# 📌 Question 3:
# Create a column "Final_Remark" based on Grade and Discipline:
# - "Outstanding" if Grade is "A" and Discipline is "Excellent"
# - "Very Good" if Grade is "B" and Discipline is "Good"
# - "Needs Improvement" otherwise

conditions3 = [
    (df['Grade'] == 'A') & (df["Discipline"] == "Excellent"),
    (df['Grade'] == 'B') & (df["Discipline"] == "Good"),
    ((df['Grade'] == 'C') | (df['Grade'] == 'D') | (df['Grade'] == 'F'))
]
values3 = [ "Outstanding","Very Good","Needs Improvement"]
df["Final_Remark"] = np.select(conditions3,values3,default='other')
df

# -------------------------------------
# 📌 Question 4:
# Create a column "Award" based on Score and Attendance:
# - "Gold" if Score >= 90 and Attendance >= 95
# - "Silver" if Score >= 80 and Attendance >= 90
# - "Bronze" if Score >= 70 and Attendance >= 85
# - "None" otherwise

conditions4 = [
    (df['Score'] >= 90) & (df['Attendance'] >= 95),
    (df['Score'] >= 80) & (df['Attendance'] >= 90),
    (df['Score'] >= 70) & (df['Attendance'] >= 85),
    (df['Score'] < 70)
]
values4 = ["Gold","Silver","Bronze","None"]
df["Award"] = np.select(conditions4,values4,default='other')
df

# -------------------------------------
# 📌 Question 5:
# Create a column "Status" based on Score and Attendance:
# - "Promoted" if Score >= 60 and Attendance >= 70
# - "Not Promoted" otherwise
conditions5 = [
    (df['Score'] >=60) & (df['Attendance']>=70),
    df['Score'] < 60

]
values5 = ["Promoted","Not Promoted"]
df['Status'] = np.select(conditions5,values5,default='other')
df
# -------------------------------------
# After completing all columns, print the final DataFrame

```
