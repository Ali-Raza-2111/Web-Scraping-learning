```python
import pandas as pd

# Sample DataFrame for practice
data = {
    'Student': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Math': [85, 90, 78, 88, 95],
    'Science': [92, 85, 80, 75, 98],
    'History': [88, 76, 94, 90, 85],
    'City': ['NYC', 'Paris', 'London', 'Tokyo', 'Berlin']
}

df = pd.DataFrame(data)
print("Original DataFrame:\n", df)

# 🧪 QUESTION 1:
# Drop the 'City' column and store the result in a new DataFrame.
city_df = df.drop('City',axis=1)

# 🧪 QUESTION 2:
# Drop the row for 'Charlie' (hint: use the index) and print the new DataFrame.
print(df.drop(index=2,axis=0))
# 🧪 QUESTION 3:
# Drop both 'Math' and 'History' columns without modifying the original DataFrame.
df.drop(columns=['Math','History'])
# 🧪 QUESTION 4:
# Drop the rows for 'Bob' and 'David' using their index values.
df.drop(index=[1,3])
# 🧪 QUESTION 5:
# Drop the 'Science' column and update the original DataFrame permanently.
df.drop(columns=['Science'],inplace=True)
df
```
