```python
import pandas as pd

employees = pd.DataFrame({
    'ID': [1, 2, 3, 4],
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Department': ['HR', 'Tech', 'Marketing', 'Finance']
})

salaries = pd.DataFrame({
    'ID': [1, 2, 3, 5],
    'Salary': [50000, 60000, 55000, 70000]
})

merged_pd = employees.merge(salaries,on = 'ID',how = 'inner')
meged_pd_full = employees.merge(salaries,on='ID',how='outer')

meged_pd_full
```

```python
import pandas as pd
import numpy as np  # For NaN values

# Employees DataFrame (some departments missing)
employees = pd.DataFrame({
    'Employee_ID': [101, 102, 103, 104],
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Department': ['HR', 'Tech', np.nan, 'Finance']  # Charlie's department missing
})

# Salaries DataFrame (some salaries missing)
salaries = pd.DataFrame({
    'Employee_ID': [101, 102, 104, 105],  # 103 missing, 105 extra
    'Salary': [50000, np.nan, 70000, 80000]  # Bob's salary missing
})

merged_outer = employees.merge(salaries,on='Employee_ID',how='outer')
merged_outer
```
