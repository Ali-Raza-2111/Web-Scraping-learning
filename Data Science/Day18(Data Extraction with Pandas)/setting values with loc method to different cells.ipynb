```python
import pandas as pd

# Create a sample DataFrame for your practice
data = {
    'Product': ['Laptop', 'Mouse', 'Keyboard', 'Monitor', 'Webcam', 'Speaker', 'Printer', 'Scanner'],
    'Price': [1200, 25, 75, 300, 50, 80, 200, 150],
    'Quantity': [10, 50, 30, 15, 40, 20, 5, 10],
    'Brand': ['Dell', 'Logitech', 'Razer', 'Samsung', 'Logitech', 'JBL', 'HP', 'Epson']
}
df_practice = pd.DataFrame(data, index=['P1', 'P2', 'P3', 'P4', 'P5', 'P6', 'P7', 'P8'])

print("Original DataFrame for practice:")
print(df_practice)
print("\n" + "="*50 + "\n")

# --- Practice Questions ---
# Try to solve these questions by writing Python code using df_practice and the .loc method.
# Then compare your answers with the solutions provided in the quiz below.

# Question 1:
# Change the 'Price' of 'Laptop' (index 'P1') to 1150
df_practice.loc['P1', 'Price'] = 1150

# Question 2:
# Update the 'Quantity' of 'Mouse' (index 'P2') to 55 and its 'Brand' to 'Microsoft'.
df_practice.loc['P2', ['Quantity', 'Brand']] = [55,'Microsoft']

# Question 3:
# Set the 'Price' of 'Keyboard' (index 'P3') and 'Monitor' (index 'P4') to 100.
df_practice.loc[ 'P3':'P4' , 'Price'] = 100

# Question 4:
# For all products with a 'Quantity' less than 15, set their 'Quantity' to 15.
df_practice.loc[df_practice['Quantity'] < 15, 'Quantity'] = 15

# Question 5:
# Add a new column 'Availability' and set its value to 'In Stock' for all products.
df_practice.loc[:, 'Availability'] = 'In Stock'

# Question 6:
# Set the 'Brand' of the 'Webcam' (index 'P5') to 'Logitech_Pro'.
df_practice.loc['P5' , 'Brand'] = 'Logitech_Pro'
df_practice

# Question 7:
# Change the 'Price' and 'Quantity' of 'Printer' (index 'P7') to 180 and 7 respectively.
df_practice.loc['P7' , ['Price','Quantity']] = [180, 7]

# Question 8:
# For products with 'Price' greater than 100, set their 'Brand' to 'Premium'.
df_practice.loc[df_practice['Price']>100 , 'Brand'] = 'Premium'

# Question 9:
# Using slicing, set the 'Quantity' for products from 'P1' to 'P3' to 20.
df_practice.loc['P1' : 'P3', 'Quantity'] = 20

# Question 10:
# Imagine you want to add a 'Discount' column and set it to 0.10 for all products initially.
df_practice.loc[:, 'Discount'] = 0.10
df_practice
```
