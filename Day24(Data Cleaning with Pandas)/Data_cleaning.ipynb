```python
import pandas as pd
```

```python
df_netflix = pd.read_csv('netflix_titles.csv')
df_netflix.head()
```

Identifying missing Data

```python
df_netflix.isnull()
```

```python
for columns in df_netflix.columns:
    percentage = df_netflix[columns].isnull().mean()
    print(columns + '  : ' + str(round(percentage*100,2)))
```

Dealing with missing Data

```python
df_netflix.dropna(subset=['director'])
```

```python
mode = ''.join(df_netflix['rating'].mode())
type(mode)

df_netflix['rating'].fillna(mode,inplace=True)
```

```python
df_netflix['duration'].fillna('0',inplace=True)
df_netflix['director'].ffill()
```

```python
df_movies = df_netflix[df_netflix['type'] == 'Movie']
df_movies['Minutes']=df_movies['duration'].str.split(expand=True)[0].astype(int)
```
