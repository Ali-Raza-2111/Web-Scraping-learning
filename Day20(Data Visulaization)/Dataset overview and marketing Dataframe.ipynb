```python
import pandas as pd
```

```python
df_population_raw = pd.read_csv('./population_total.csv')
df_population_raw
```

```python
df_population_raw.dropna(inplace=True)
```

```python
df_pivot = df_population_raw.pivot(index='year',columns='country',values='population')
df_pivot
```

```python
df_pivot = df_pivot[['Afghanistan','Albania','Algeria','Pakistan']]
```

```python
df_pivot
```

```python
df_pivot.plot(kind='line',xlabel='Years',title='Population Chart',ylabel='Population',figsize=(8,4))
```

```python
df_pivot_2020 = df_pivot[df_pivot.index.isin([2020])]
df_pivot_2020
```

```python
df_pivot_2020 = df_pivot_2020.T
```

```python
df_pivot_2020.plot(kind='bar',xlabel='Years',title='Population Chart',color='orange')
```

```python
df_pivot_Sample= df_pivot[df_pivot.index.isin([1980,1990,2000,2010,2020])]
df_pivot_Sample
```

```python
df_pivot_Sample.plot(kind='bar',xlabel='Years',title='Population Chart',ylabel='Population')
```

```python
df_pivot_2020.rename(columns={2020:"2020"},inplace=True)
```

```python
df_pivot_2020.plot(kind='pie',y='2020',title='Population Chart')
```
